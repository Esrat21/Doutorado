Como boa parte das operaçoes em deep learningocorre por por meio de operaçoes de matrizes, e as matrizes tem uma regra muito rígida para tamanhos e formatos, causam o mais comum dos erros quando executamso deep learning, "Shape mismatch"

```python
# Shapes need to be in the right way  
tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]], dtype=torch.float32)

tensor_B = torch.tensor([[7, 10],
                         [8, 11], 
                         [9, 12]], dtype=torch.float32)

torch.matmul(tensor_A, tensor_B) # erro por causa da dimensão dos tensores, para multiplicação de matrizes, o número de colunas do primeiro tensor deve ser igual ao número de linhas do segundo tensor
```

Para poder multiplicar dois tensores, (tensorA) e (tensorB), precisamos ajustar suas dimensões adequando as regras de multiplicaçao de matrizes
