## 2️⃣ Tensores: A Base de Tudo no PyTorch

**O que é um tensor?**
- Um **tensor** é uma generalization de matrizes para N dimensões
- Um vetor é um tensor de 1 dimensão
- Uma matriz é um tensor de 2 dimensões
- Uma imagem colorida é um tensor de 3 dimensões (altura, largura, canais de cor)
- Um batch de imagens é um tensor de 4 dimensões

Os tensores são a estrutura fundamental no PyTorch - **tudo** são tensores!

````python
# =============================================
# CRIANDO TENSORES
# =============================================

# Tensor a partir de uma lista Python
# torch.tensor() converte dados Python em tensores PyTorch
lista = [1.0, 2.0, 3.0, 4.0]
tensor1 = torch.tensor(lista)
print(f"Tensor de lista: {tensor1}")
print(f"Tipo: {type(tensor1)}, Shape: {tensor1.shape}, dtype: {tensor1.dtype}")

# Tensor de zeros (útil para inicializar pesos)
# torch.zeros() cria um tensor preenchido com zeros
zeros = torch.zeros(3, 4)  # 3 linhas, 4 colunas = matriz 3x4
print(f"\nTensor de zeros (3x4):\n{zeros}")

# Tensor de uns (útil para mascaras ou inicialização)
# torch.ones() cria um tensor preenchido com uns
uns = torch.ones(2, 3)
print(f"\nTensor de uns (2x3):\n{uns}")

# Tensor com valores aleatórios (distribuição normal)
# torch.randn() gera valores de uma distribuição normal padrão (média=0, desvio=1)
aleatorio = torch.randn(3, 3)  # matriz 3x3 com valores aleatórios
print(f"\nTensor aleatório (3x3):\n{aleatorio}")

# Tensor a partir de arrays NumPy
# Isso permite integração fácil com outras bibliotecas
array_np = np.array([[1, 2, 3], [4, 5, 6]])
tensor_np = torch.from_numpy(array_np)
print(f"\nTensor de NumPy array:\n{tensor_np}")
print(f"Shape: {tensor_np.shape}")
````

````python
# =============================================
# OPERAÇÕES BÁSICAS COM TENSORES
# =============================================

# Criar dois tensores para演示 operações
a = torch.tensor([1.0, 2.0, 3.0])
b = torch.tensor([4.0, 5.0, 6.0])

# Adição
soma = a + b
print(f"{a} + {b} = {soma}")

# Multiplicação elemento por elemento
mult = a * b
print(f"{a} * {b} = {mult}")

# Multiplicação por escalar
escalar = 2 * a
print(f"2 * {a} = {escalar}")

# Producto escalar (dot product)
dot = torch.dot(a, b)  # 1*4 + 2*5 + 3*6 = 32
print(f"Dot product: {a} · {b} = {dot}")

# Soma de todos os elementos
soma_total = a.sum()
print(f"Soma de {a} = {soma_total}")

# reshaping: mudar a forma do tensor
tensor_grande = torch.randn(12)  # 12 elementos
print(f"\nTensor original (12 elementos): {tensor_grande.shape}")
reshaped = tensor_grande.view(3, 4)  # transforma em matriz 3x4
print(f"Reshaped (3x4):\n{reshaped}")

# Flattening: voltar para 1D
flat = reshaped.view(-1)  # -1 significa "calcular automaticamente"
print(f"Flat: {flat.shape}")
````

## 3️⃣ Autograd: Cálculo Automático de Gradientes

**O que é autograd?**
- `torch.autograd` é o sistema de **diferenciação automática** do PyTorch
- Ele constrói automaticamente o **grafo computacional** de cada operação
- Quando chamamos `.backward()`, ele calcula todos os gradientes automaticamente
- Isso é essencial para o **Backpropagation** (propagação reversa do erro)

**Por que isso é importante?**
- Sem autograd, teríamos que calcular todas as derivadas manualmente (imagine uma rede com milhões de parâmetros!)
- Com autograd, o PyTorch faz isso automaticamente usando a **regra da cadeia**

````python
# =============================================
# ENTENDENDO O AUTOGRAD
# =============================================

# Para habilitar o autograd, criamos tensores com requires_grad=True
# Isso diz ao PyTorch para "rastrear" todas as operações com esse tensor
x = torch.tensor([2.0, 3.0], requires_grad=True)  # input
y = torch.tensor([4.0, 5.0], requires_grad=True)  # input

# Vamos fazer uma operação: z = x^2 + y^2 + 2*x*y
z = x**2 + y**2 + 2*x*y

# Isso é equivalente a: z = (x + y)^2

print(f"x = {x}")
print(f"y = {y}")
print(f"z = x² + y² + 2xy = {z}")

# Para calcular os gradientes, chamamos .backward() em z
# Mas z é um vetor, então precisamos dar um gradiente inicial (como se fosse um peso)
z.sum().backward()  # z.sum() porque backward() espera um escalar

# Agora x.grad e y.grad contém os gradientes de z em relação a x e y
# Para z = (x+y)², a derivada de z em relação a x é: dz/dx = 2*(x+y)
# Para z = (2+3)² = 25, dz/dx = 2*(2+3) = 10

print(f"\nGradiente de z em relação a x: dz/dx = {x.grad}")
print(f"Gradiente de z em relação a y: dz/dy = {y.grad}")
print(f"\nVerificação: dz/dx deveria ser 2*(x+y) = 2*(2+3) = 10 ✓")
````

## 4️⃣ Criar o Dataset

Vamos criar um **dataset sintético** para classificação. Será pontos em 2D que pertencem a duas classes diferentes (0 ou 1).

**Problema**: Dado um ponto (x, y), prever se pertence à classe 0 ou classe 1.

````python
# =============================================
# CRIAR DATASET SINTÉTICO PARA CLASSIFICAÇÃO
# =============================================

# Vamos criar dados que formam duas "nuvens" de pontos
# Classe 0: pontos centrados em (-1, -1)
# Classe 1: pontos centrados em (1, 1)

np.random.seed(42)  # Semente para reprodutibilidade (mesmos resultados sempre)
torch.manual_seed(42)  # Semente também para PyTorch

n_amostras = 100  # Número de amostras por classe

# Classe 0: distribuição normal em torno de (-1, -1)
# media = [-1, -1], desvio padrão = 0.5
classe0 = torch.randn(n_amostras, 2) * 0.5 + torch.tensor([-1.0, -1.0])
labels0 = torch.zeros(n_amostras, dtype=torch.long)  # Label 0

# Classe 1: distribuição normal em torno de (1, 1)
# media = [1, 1], desvio padrão = 0.5
classe1 = torch.randn(n_amostras, 2) * 0.5 + torch.tensor([1.0, 1.0])
labels1 = torch.ones(n_amostras, dtype=torch.long)  # Label 1

# Concatenar: axis=0 significa empilhar verticalmente
X = torch.cat([classe0, classe1], dim=0)  # shape: (200, 2) - 200 amostras, 2 features
y = torch.cat([labels0, labels1], dim=0)  # shape: (200,) - 200 labels

print(f"Dataset criado!")
print(f"X (features): shape = {X.shape}")
print(f"y (labels): shape = {y.shape}")
print(f"Classes: {y.unique()} (0 e 1)")

# Visualizar os dados
plt.figure(figsize=(8, 6))
plt.scatter(X[:n_amostras, 0], X[:n_amostras, 1], c='blue', label='Classe 0', alpha=0.7)
plt.scatter(X[n_amostras:, 0], X[n_amostras:, 1], c='red', label='Classe 1', alpha=0.7)
plt.xlabel('Feature 1 (x₁)')
plt.ylabel('Feature 2 (x₂)')
plt.title('Dataset de Classificação Binária')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
````

## 5️⃣ Definir o Modelo Neural

Agora vamos criar nossa **rede neural**! Usaremos `torch.nn.Module` como classe base.

**Arquitetura da nossa rede:**
- **Input Layer**: 2 features (x₁, x₂)
- **Hidden Layer 1**: 16 neurônios + ReLU (ativação)
- **Hidden Layer 2**: 16 neurônios + ReLU (ativação)
- **Output Layer**: 2 classes (saída para CrossEntropyLoss)

**Conceitos importantes:**
- `nn.Linear(in_features, out_features)`: camada totalmente conectada (y = Wx + b)
- `nn.ReLU()`: função de ativação que introduce não-linearidade
- `forward()`: método que define como os dados fluem pela rede

````python
# =============================================
# DEFINIR A ARQUITETURA DA REDE NEURAL
# =============================================

class RedeNeuralSimples(nn.Module):
    """
    Rede Neural para classificação binária.
    
    Arquitetura:
    - Input: 2 features (x₁, x₂)
    - Hidden 1: 16 neurônios + ReLU
    - Hidden 2: 16 neurônios + ReLU
    - Output: 2 classes ( logits para CrossEntropyLoss)
    """
    
    def __init__(self):
        # super().__init__() é essencial! Chamando o inicializador da classe pai (nn.Module)
        super().__init__()
        
        # Definir as camadas da rede
        # nn.Linear: y = W*x + b (transformação linear)
        # Parâmetros: (tam_entrada, tam_saída)
        self.camada_oculta1 = nn.Linear(2, 16)   # 2 -> 16
        self.camada_oculta2 = nn.Linear(16, 16)  # 16 -> 16
        self.camada_saida = nn.Linear(16, 2)     # 16 -> 2
        
        # Funções de ativação (não têm parâmetros aprendíveis)
        self.ativacao = nn.ReLU()
        
    def forward(self, x):
        """
        Define como os dados passam pela rede.
        Você DEVE implementar este método!
        
        Args:
            x: tensor de entrada (batch_size, 2)
        Returns:
            tensor de saída (batch_size, 2)
        """
        # Passagem pela primeira camada oculta + ativação
        x = self.ativacao(self.camada_oculta1(x))
        
        # Passagem pela segunda camada oculta + ativação
        x = self.ativacao(self.camada_oculta2(x))
        
        # Passagem pela camada de saída (SEM ativação aqui!)
        # CrossEntropyLoss espera "logits" (valores antes do softmax)
        x = self.camada_saida(x)
        
        return x

# Instanciar o modelo
modelo = RedeNeuralSimples()

# Ver a arquitetura da rede
print("🏗️  Arquitetura da Rede Neural:")
print(modelo)

# Listar todos os parâmetros aprendíveis
print("\n📋 Parâmetros da rede (pesos e biases):")
for nome, param in modelo.named_parameters():
    print(f"  {nome}: shape = {param.shape}")
````

## 6️⃣ Definir Loss Function e Otimizador

### Loss Function (Função de Perda)
- **CrossEntropyLoss**: Mede o quão "errada" está a predictions do modelo
- Para classificação com 2 classes, é a função mais comum
- Valores menores = melhor (0 = predição perfeita)

### Otimizador (Optimizer)
- **SGD (Stochastic Gradient Descent)**: O otimizador mais básico
  - `lr` (learning rate): tamanho do passo de cada atualização
  - `momentum`:惯性，帮助加速收敛
- **Adam**: Otimizador mais moderno e geralmente mais rápido
  - Combina ideias de momentum e adaptive learning rates

````python
# =============================================
# DEFINIR LOSS FUNCTION E OTIMIZADOR
# =============================================

# Loss Function: mede o erro entre predições e valores reais
# CrossEntropyLoss combina LogSoftmax + NLLLoss em uma só função
criterion = nn.CrossEntropyLoss()

# Otimizador: atualiza os pesos da rede para minimizar o loss
# SGD = Stochastic Gradient Descent
# - lr (learning rate): quão grande é cada passo de atualização
# - momentum: ajuda a acelerar a convergência e escapar de mínimos locais
learning_rate = 0.1
momentum = 0.9

optimizer = optim.SGD(modelo.parameters(), lr=learning_rate, momentum=momentum)

print("⚙️  Configurações de treinamento:")
print(f"   Loss Function: CrossEntropyLoss")
print(f"   Otimizador: SGD")
print(f"   Learning Rate: {learning_rate}")
print(f"   Momentum: {momentum}")
print(f"\n📊 Número total de parâmetros a serem treinados: {sum(p.numel() for p in modelo.parameters())}")
````

---

## 7️⃣ Loop de Treinamento

Este é o coração do treinamento de qualquer modelo em PyTorch!

### O algoritmo completo (1 época = 1 passagem por todos os dados):

1. **Forward Pass**: passar os dados pela rede → obter predições
2. **Calcular Loss**: comparar predições com valores reais → obter erro
3. **Backward Pass**: calcular gradientes do loss em relação aos pesos
4. **Otimizar**: atualizar os pesos na direção que reduz o erro
5. **Zero Grad**: limpar os gradientes para a próxima iteração

**Repetir** por N épocas até o loss convergir (diminuir o suficiente)

````python
# =============================================
# LOOP DE TREINAMENTO
# =============================================

# Hiperparâmetros de treinamento
n_epocas = 100  # Número de vezes que passamos por TODO o dataset

# Listas para guardar o histórico de loss (para plotar depois)
historico_loss = []

print("🚀 Iniciando treinamento...")
print("=" * 50)

for epoca in range(n_epocas):
    # ===== 1. FORWARD PASS =====
    # Passar os dados pela rede neural
    # X: tensor com shape (200, 2) - 200 amostras, 2 features
    # output: tensor com shape (200, 2) - 200 amostras, 2 classes (logits)
    predictions = modelo(X)
    
    # Calcular o loss (erro)
    # y: tensor com shape (200,) - os labels verdadeiros
    # predictions: shape (200, 2)
    loss = criterion(predictions, y)
    
    # ===== 2. BACKWARD PASS =====
    # Calcular os gradientes automaticamente via autograd
    # Isso é o "backpropagation" - propaga o erro para trás na rede
    loss.backward()
    
    # ===== 3. OTIMIZAR (atualizar pesos) =====
    # Atualizar cada parâmetro na direção que reduz o loss
    optimizer.step()
    
    # ===== 4. ZERO GRAD =====
    # IMPORTANTÍSSIMO: zerar os gradientes antes da próxima iteração
    # Se não fizermos isso, os gradientes ACUMULAM e o treinamento explode!
    optimizer.zero_grad()
    
    # Salvar o loss para plotar depois
    historico_loss.append(loss.item())
    
    # A cada 10 épocas, mostrar o progresso
    if (epoca + 1) % 10 == 0:
        # Calcular a acurácia manualmente
        # .max(1) retorna (valores_maximos, indices) no eixo 1 (as classes)
        _, predicted_classes = torch.max(predictions, dim=1)
        acuracia = (predicted_classes == y).float().mean().item()
        
        print(f"Época [{epoca+1:3d}/{n_epocas}] | Loss: {loss.item():.4f} | Acurácia: {acuracia*100:.1f}%")

print("=" * 50)
print("✅ Treinamento concluído!")
````

## 8️⃣ Avaliar e Visualizar Resultados

Vamos visualizar:
1. A curva de aprendizado (como o loss diminuiu ao longo do tempo)
2. A fronteira de decisão aprendida pela rede neural

````python
# =============================================
# VISUALIZAR A CURVA DE APRENDIZADO
# =============================================

plt.figure(figsize=(10, 4))

# Plot do loss ao longo das épocas
plt.subplot(1, 2, 1)
plt.plot(historico_loss, 'b-', linewidth=2)
plt.xlabel('Época')
plt.ylabel('Loss')
plt.title('Curva de Aprendizado')
plt.grid(True, alpha=0.3)

# Plot da acurácia (calculada a partir do histórico)
# Para fazer isso, precisamos recalcular durante o treinamento
# Por agora, apenas mostre o loss final
plt.subplot(1, 2, 2)
plt.plot(historico_loss, 'b-', linewidth=2)
plt.xlabel('Época')
plt.ylabel('Loss')
plt.title(f'Loss Final: {historico_loss[-1]:.4f}')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\n📈 Loss inicial: {historico_loss[0]:.4f}")
print(f"📉 Loss final: {historico_loss[-1]:.4f}")
print(f"📊 Redução do loss: {(1 - historico_loss[-1]/historico_loss[0])*100:.1f}%")
````

````python
# =============================================
# VISUALIZAR A FRONTEIRA DE DECISÃO
# =============================================

# A fronteira de decisão mostra onde a rede "decide" entre uma classe e outra

# Criar uma meshgrid (grade) de pontos no espaço 2D
x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 100), np.linspace(y_min, y_max, 100))

# Para cada ponto na grade, fazer uma predição
# Flatten para usar no modelo
grid_points = torch.tensor(np.c_[xx.ravel(), yy.ravel()], dtype=torch.float32)

# Forward pass com o modelo treinado
with torch.no_grad():  # Sem calcular gradientes (só predição)
    Z = modelo(grid_points)
    # Pegar a classe com maior score (argmax)
    Z = torch.argmax(Z, dim=1).numpy()

Z = Z.reshape(xx.shape)

# Plotar
plt.figure(figsize=(10, 8))

# Plot da fronteira de decisão (regiões de cor)
plt.contourf(xx, yy, Z, alpha=0.3, cmap='coolwarm')

# Plot dos pontos de dados
plt.scatter(X[:n_amostras, 0], X[:n_amostras, 1], c='blue', label='Classe 0', edgecolors='k', s=50)
plt.scatter(X[n_amostras:, 0], X[n_amostras:, 1], c='red', label='Classe 1', edgecolors='k', s=50)

plt.xlabel('Feature 1 (x₁)')
plt.ylabel('Feature 2 (x₂)')
plt.title('Fronteira de Decisão da Rede Neural')
plt.legend()
plt.colorbar(label='Classe predita')
plt.grid(True, alpha=0.3)
plt.show()

# Calcular acurácia final
with torch.no_grad():
    predictions = modelo(X)
    _, predicted = torch.max(predictions, dim=1)
    acuracia = (predicted == y).float().mean()
    print(f"\n🎯 Acurácia final no dataset: {acuracia.item()*100:.1f}%")
````

### 🔑 O loop de treinamento básico:

```
for época in range(N):
    1. predictions = model(X)           # Forward
    2. loss = criterion(predictions, y) # Calcular erro
    3. loss.backward()                   # Backward (gradientes)
    4. optimizer.step()                  # Atualizar pesos
    5. optimizer.zero_grad()             # Zerar gradientes
```
