```python
import pygame
import random
import numpy as np
from enum import Enum
from collections import namedtuple, deque

import torch
import torch.nn as nn
import torch.optim as optim
```

```python
pygame.init()
screen = pygame.display.set_mode((600, 600))

clock = pygame.time.Clock()
running = True

GRID_SIZE = 20
CELL_SIZE = 30
GRID_WIDTH = 600 // GRID_SIZE
GRID_HEIGHT = 600 // GRID_SIZE
GRID_COLOR = (40, 40, 40)
SNAKE_COLOR = (0, 255, 0)

while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    screen.fill((0, 0, 0))
    
    snake_body = [(GRID_WIDTH // 2, GRID_HEIGHT // 2)]
    for segment in snake_body:
        pygame.draw.rect(screen, SNAKE_COLOR, (segment[0] * CELL_SIZE, segment[1] * CELL_SIZE, CELL_SIZE, CELL_SIZE))

    pygame.display.flip()
    clock.tick(60)

pygame.quit()
```

```python
class snake:
    def __init__(self):
        self.body = [(GRID_WIDTH // 2, GRID_HEIGHT // 2)]
        self.direction = (0, -1)  # começa indo para cima
        self.grow = False

    def move(self):
        head_x, head_y = self.body[0]
        dir_x, dir_y = self.direction
        new_head = (head_x + dir_x, head_y + dir_y)

        if self.grow:
            self.body.insert(0, new_head)
            self.grow = False
        else:
            self.body.insert(0, new_head)
            self.body.pop()
```
